# Dating App Culture Through User Reviews

## Motivation & problem

Dating apps have become one of the main ways people my age meet partners. Many of my friends use them, and several found their current partners through one. At the same time, Instagram is full of memes contrasting "dating today vs. dating in the past," usually suggesting that modern dating has become more frustrating, shallow, or exhausting.

These memes reflect a popular perception, but they are anecdotal. It is unclear whether users' actual experiences with dating apps have gotten worse, and if so, what is driving the frustration. Is it fake profiles, paywalled features, ghosting, safety concerns, or simply not getting matches?

This project uses user reviews of Tinder, Bumble, and Hinge to study that frustration through the apps themselves. Reviews are mostly about features, matching, and payments, so I treat them as evidence of how app design shapes the dating experience, not as a direct measure of dating culture as a whole.

My guiding question is: **How do user complaints differ across Tinder, Bumble, and Hinge, and how do those differences relate to each app's design choices?** For example, does Bumble's women-message-first design lead to fewer safety complaints? As a secondary question, I will look at how these complaints changed between 2017 and 2022.

**Whose experiences this covers:** The reviews have no age information, so I cannot narrow the group by age. Instead, I narrow it to English-language Google Play reviewers who describe a specific part of the app experience.

I hope to learn which design decisions frustrate users most. These findings could help app designers decide what to fix, and help users see whether their frustrations are widely shared.

## Data

I plan to use the **Dating Apps Reviews 2017–2022 (all Regions)** dataset on Kaggle: https://www.kaggle.com/datasets/sidharthkriplani/datingappreviews

**What it contains:** Google Play Store reviews of Tinder, Bumble, and Hinge. Each row includes the reviewer's name, the review text, a 1–5 star rating, the number of "thumbs up" the review received, the date, and the app. Other analyses of this dataset report about 682,000 reviews, roughly 77% of them for Tinder.

**License / terms:** [CCO: Public Domain] The reviews were originally posted publicly on the Google Play Store.

**Why it fits:** The reviews are written by real users in their own words, so they capture the frustrations that memes only describe. The dataset spans several years, which lets me look at change over time, and it covers three apps with different designs, which allows a comparison.

**Ethical considerations:**

- **Consent:** Reviews are public, but users did not write them expecting to be studied. I will remove reviewer names before analysis.
- **Quoting:** Quoting full reviews could let someone search for and identify the author. I will paraphrase or quote only short fragments.
- **Sensitive content:** Some reviews may describe harassment or include personal details. I will avoid highlighting identifiable personal stories.
- **Representativeness:** People who leave reviews tend to be unusually happy or unhappy, and the data covers only Android users. My findings will describe reviewers, not all dating app users.
- **No age data:** The dataset has no age information, so I will not claim findings are about Gen Z or any age group. My conclusions will be about how app design relates to reviewers' complaints.

### Planned analysis

1. **Clean and sample:** Remove reviewer names, keep English-language reviews, and draw a sample balanced by app and year.
2. **Define complaint categories:** Start with six categories: fake profiles and bots, payments and paywalls, matching and visibility, safety and harassment, conversation quality (e.g. ghosting), and technical bugs. I will hand-label about 300 reviews to refine these definitions.
3. **Label the rest:** Apply the categories to the full sample with keyword rules or topic modeling, and use star ratings plus a sentiment tool (e.g. VADER) to measure tone.
4. **Compare:** Calculate the share of each complaint type per app and per year, then connect differences to each app's design features.
5. **Validate:** Compare automatic labels against my hand labels and report how often they agree.

## Unknowns & dependencies

- **License:** If the Kaggle license is unclear or restrictive, I may need to limit how I share the data in my repo (for example, sharing only my processing code, not the raw file).
- **Dataset size:** Around 680,000 reviews may be slow to process in a notebook. I plan to sample if needed, balanced by app and year.
- **Imbalance across apps:** Tinder makes up most of the reviews, so comparisons between apps will need balanced samples or normalized counts.
- **Validating the methods:** Automated sentiment and topic tools can misread sarcasm and slang. I will hand-check a sample of labels, which will take time I need to budget for.
- **Time coverage:** The data ends in 2022, so it cannot speak to the most recent dating app trends.
- **No age data:** I cannot confirm how many reviewers are Gen Z, which is why my question focuses on app design rather than age groups.
